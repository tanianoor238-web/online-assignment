```python
import csv
```

```python
open_file=open('student data (2).csv',mode='r')
open_file
```

```python
reader=csv.reader(open_file)
reader
```

```python
data=list(reader)
```

```python
data
```

```python
header=data[0]
```

```python
header
```

```python
student_data=data[1:]
```

```python
student_data
```

```python
header.append("Grade")
```

```python
header
```

```python
for student in student_data:
    percentage=float(student[5])
    if percentage>=80:
        grade="A"
    elif percentage>=70 :
        grade="B"
    elif percentage>=60:
        grade="C"
    elif percentage>=50:
        grade="D"
    else:
        grade="F"
    student.append(grade)
```
